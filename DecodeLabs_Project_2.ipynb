# **Project : Expense Tracker**

**Track user expenses using accumulator pattern**

**PROBLEM STATEMENT:**
Baat simple hai — jab koi banda apna paise track nahi karta, end of month mein pata chalta hai "yaar paisa kahan gaya?" Yahi problem hai. Hume ek Expense Tracker banana hai jahan user apne kharchay enter karta jaaye, aur program sab add karta rahe aur final total dikhaye.

**OBJECTIVE:**
i. User baar baar expense amount enter karega (100, 50, 200...)
ii. Program har entry loop mein add karta jaayega accumulator total mein
iii. Jab user done likhe → program total spend dikhaye aur band ho jaaye
iv. Galat input (like "abc") → crash nahi, warning do aur continue karo

**APPROACH — Step by Step:** Step 1 → total = 0  (loop se BAHAR — Quality Standard #2)
Step 2 → while loop shuru — user se input lo
Step 3 → agar "done" type kiya → loop tod do (Kill Switch — QS #4)
Step 4 → try/except se ValueError pakdo (Defense — QS #3)
Step 5 → valid number mila → total = total + expense (Accumulator)
Step 6 → Loop khatam → Final Total print karo

```python

# SECTION: Startup Banner
# WHAT: Just a clean welcome screen for the user
# WHY: Professional feel + user knows what to do
print("=" * 45)
print("       💸 DECODELABS EXPENSE TRACKER 💸      ")
print("=" * 45)
print("➡  Enter expense amounts one by one.")
print("➡  Type  'done'  when you're finished.\n")

# SECTION: State Initialization
# WHAT: Initialize total BEFORE the loop starts
# WHY: If we put this inside the loop, it would reset to 0
# every single time — we'd lose all previous expenses!
# Jaise scoreboard innings start hone se pehle 0 hota hai

total = 0.0             # This is our ACCUMULATOR — it grows with each expense
transaction_count = 0   # Counting how many valid expenses were entered

# SECTION: Main Input Loop
# WHAT: Keeps asking for input until user types 'done'
# WHY: while True gives us an infinite loop that only we control
# This handles 5, 10, 100+ transactions — no limit, no crash
while True:
    # Get raw input from user — always a string at this point
    user_input = input("Enter expense amount (or 'done' to finish): ").strip()

    # SECTION: Kill Switch
    # WHAT: Check if user wants to stop
    # WHY: .lower() handles "Done", "DONE", "done" — all work
    if user_input.lower() == "done":
        print("\n🛑 Kill switch activated! Calculating your total...\n")
        break  # Exit the loop — go straight to final total

    # SECTION: Input Defense
    # WHAT: Try to convert input to float, catch bad inputs
    # WHY: If user types "abc" or "hello", float() will crash
    #with ValueError — we CATCH it instead of dying
    # Jaise fielder ball drop kare, game rukta nahi — chalta rehta hai
    try:
        expense = float(user_input)  # Convert string → number

        # Extra check: negative expenses don't make sense
        if expense < 0:
            print("⚠️  Negative expense? Bhai yeh bank robbery nahi hai 😄 Try again.\n")
            continue  # Skip this, go back to top of loop

        # SECTION: Accumulator Logic — THE CORE OF THIS TASK
        # WHAT: Add new expense to running total
        # WHY: total = total + expense is the accumulator pattern
        #      Har ball ke runs scoreboard pe add hote hain — same thing
        total = total + expense          # Accumulator: total grows with each entry
        transaction_count += 1           # Count valid transactions
        # Live feedback — user sees running total after each entry
        print(f"Added: Rs. {expense:.2f}  |  Running Total: Rs. {total:.2f}\n")

    except ValueError:
        # This runs ONLY when float(user_input) fails
        # e.g., user typed "fifty" or "abc" or left it empty
        print(f"'{user_input}' is not a valid number. Skip karo aur dobara try karo!\n")
        # Loop continues automatically — no crash, no exit

# SECTION: Final Report
# WHAT: After kill switch, display the complete expense summary
# WHY: This is the whole point — total kharch batana
print("=" * 45)
print("FINAL EXPENSE REPORT")
print("=" * 45)
print(f"  Total Transactions : {transaction_count}")
print(f"  Total Amount Spent : Rs. {total:.2f}")
print("=" * 45)

# Handle edge case: user typed 'done' without entering anything
if transaction_count == 0:
    print("No expenses were recorded. Mast banda hai — kharcha hi nahi kiya!")
else:
    print(f"Average per expense: Rs. {total / transaction_count:.2f}")
```
